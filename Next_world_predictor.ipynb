### Importing library

```python
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, LSTM, Dense
import numpy as np
from tensorflow.keras.utils import to_categorical
from sklearn.model_selection import train_test_split
```

### Tokenization


```python
tokenizer = Tokenizer()
tokenizer.fit_on_texts([data])
```

### Seeing index

```python
len(tokenizer.word_index)
```

### Sentences

```python
sentences = data.split('.')
```

### Converting into sequences

```python
tokenizer = Tokenizer()
tokenizer.fit_on_texts(sentences)

input_sequences = []

for sentence in sentences:
    token_list = tokenizer.texts_to_sequences([sentence])[0]

    for i in range(1, len(token_list)):
        sequence = token_list[:i+1]
        input_sequences.append(sequence)
```

### Checking max sequence length

```python
sequences = tokenizer.texts_to_sequences(sentences)
```

```python
max_sequence_len = max([len(x) for x in input_sequences])

print(max_sequence_len)
```

### Padding sequences

```python
pad_sequences_input = pad_sequences(input_sequences, maxlen=max_sequence_len, padding='pre')
```

```python
pad_sequences_input
```

### Seperating input and output

```python
X = pad_sequences_input[:,:-1]
y = pad_sequences_input[:,-1]
```

### OHE output

```python
y = to_categorical(y, num_classes=len(tokenizer.word_index)+1)
```

### Creating model

```python
model = Sequential()
model.add(Embedding(len(tokenizer.word_index) + 1, 100, input_length=max_sequence_len-1))
model.add(LSTM(150))
model.add(Dense(len(tokenizer.word_index) + 1, activation='softmax'))
model.compile(loss='categorical_crossentropy', optimizer='adam', metrics=['accuracy'])
```

### Fit model

```python
model.fit(X,y,epochs=10)
```

### Predicting

```python
text = "In a small town"
for i in range(5):
  ### Tokenize
  token_text = tokenizer.texts_to_sequences([text])[0]
  ### Padding
  pad_test = pad_sequences([token_text], maxlen=max_sequence_len-1, padding='pre')
  ### Predict
  predict_x=model.predict(pad_test)
  pos = np.argmax(predict_x)
  for word,index in tokenizer.word_index.items():
    if index == pos:
      text = text + ' ' + word
      print(text)
```
